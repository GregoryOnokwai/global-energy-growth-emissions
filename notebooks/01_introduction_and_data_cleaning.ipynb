<h2 style="color: #1F4E79;">2. The Business Value</h2>

<ul>
  <li style="margin-bottom: 10px;">
    &#128176; <b>Capital allocation.</b> Clustering and the sustainability index let a funder <b>rank and segment countries</b>,
    for example <i>"high-need, low-access"</i> versus <i>"high-emitting industrial"</i>.
    That shows <b>where a dollar of financing does the most good</b>. This is the <b>main commercial use case</b>
    for development banks, climate funds, and renewable developers screening markets.
  </li>
  <li style="margin-bottom: 10px;">
    &#127963; <b>Policy targeting.</b> Models that show which energy variables are most associated with <b>GDP</b> and <b>CO<sub>2</sub></b>
    tell a ministry <b>which levers to prioritize</b>, such as clean cooking access, fossil-generation share, or energy intensity.
    Used carefully, they also support <b>cost-benefit cases</b> for transition programs.
  </li>
  <li style="margin-bottom: 10px;">
    &#128680; <b>Early warning and monitoring.</b> Forecasts with <b>uncertainty bands to 2030</b>, plus an <b>off-track flag</b>,
    turn SDG reporting from a <i>backward-looking audit</i> into something teams can <b>act on</b>.
    A live dashboard keeps the index current without re-running the analysis by hand.
  </li>
  <li style="margin-bottom: 10px;">
    &#128202; <b>Risk and benchmarking.</b> <b>Peer-group comparisons</b> help investors and analysts judge whether a country's
    energy transition is <b>credible</b>, <b>stalling</b>, or <b>ahead</b> of similar economies.
    That feeds <b>climate-risk</b> and <b>country-risk</b> assessments.
  </li>
  <li style="margin-bottom: 10px;">
    &#9989; <b>Credibility.</b> Hypothesis tests, causal checks, and honest uncertainty make the findings
    <b>defensible in front of a board or a ministry</b>, which pure correlation charts are not.
  </li>
</ul>

<h2 style="color: #1F4E79;">3. Honest Limits</h2>

<ul>
  <li style="margin-bottom: 10px;">
    <b>Unit of analysis is the country-year.</b> The data suits <b>macro-level strategy and prioritization</b>,
    <u>not</u> project-level or utility-level decisions.
  </li>
  <li style="margin-bottom: 10px;">
    <b>Associations, not proof.</b> Energy use and GDP rise together partly because <i>richer countries can afford more of both</i>.
    Fixed-effects regression, Granger tests, and quasi-experimental designs <b>narrow this gap but don't close it</b>,
    so findings should <b>inform decisions rather than dictate them</b>.
  </li>
  <li style="margin-bottom: 10px;">
    <b>Data quality varies.</b> Gaps in the data and uneven country coverage mean
    <b>the least reliable numbers are often for the least-developed countries</b>, which are the ones most in need.
    <i>This is why the cleaning stage below matters.</i>
  </li>
</ul>

<blockquote style="border-left: 5px solid #1F4E79; padding: 10px 14px; margin: 16px 0;">
  <b>&#128204; In one sentence:</b> The value is a <b>defensible, repeatable way</b> to decide
  <b>where to intervene</b>, <b>how much progress to expect</b>, and <b>who is falling behind</b>,
  as long as no one over-reads the correlations.
</blockquote>

<h2 style="color: #1F4E79;">4. Data Cleaning: What This Notebook Does</h2>

<p>Before any analysis can answer the questions above, the raw data must be reliable. This notebook:</p>

<ol>
  <li><b>Loads</b> the raw dataset and inspects its shape, data types, and coverage by country and year.</li>
  <li><b>Descriptive statistics and missingness</b> by column, country, and year to see where the gaps are concentrated.</li>
  <li><b>Skewness analyses</b> to measure asymmetry of data distribution and show which direction outliers lie.</li>
  <li><b>Standardize column names</b> by renaming using snake case, removing hidden spaces and line breaks, ensuring a consistent style.</li>
  <li><b>Fixes data types</b> by checking and confirming columns contain the correct data type; clean data types.</li>
  <li><b>Duplicates and panel structure</b> to find and handle duplicates, and ensure panel balancing when necessary.</li>
  <li><b>Validity rules</b> through physical (what the system can handle) and logical limits (makes sense for business or real-world scenario).</li>
  <li><b>Missing value audit</b> to measure, map and understand gaps where data should be.</li>
  <li><b>Fills gaps</b> using facts and definitions, not estimates. Structural fixes using deterministic methods (eliminates randomness), not statistical.   <li><b>Country data-coverage score</b> allows reporting of results with and without low-coverage countries.</li>

  <li><b>What still needs care</b> </li>

<h3 style="color: #1F4E79;">4.2. Descriptive statistics and missingness: </h3>

```python
def summarize_data(df):
    """Summarize statistics and missing values"""
    
    # Basic statistics
    print("=== BASIC STATISTICS ===")
    print(df.describe())
    
    # Missing values
    print("\n=== MISSING VALUES ===")
    missing = df.isnull().sum()          
    missing_pct = (missing / len(df)) * 100          # Converts the count to percentages
    missing_df = pd.DataFrame({'Missing Count': missing, 'Missing %': missing_pct}) # Combines the two series into a dataframe missing_df
    print(missing_df[missing_df['Missing Count'] > 0])          # Shows columns with at least one missing value
    
    return missing_df          # Sends the full missing value table when function is called
# Run EDA
eda_results = summarize_data(df)
```

<h4 style="color: #1F4E79;">Missing value visualizations:  </h4>

```python
"""
Missing-value visualizations for the dataset.
Figure 1: missing % per column, plotted as a ranked horizontal bar chart (from eda_results).
Figure 2: missing % per column and year, plotted as a heatmap (from df). It shows whether
          gaps cluster in certain years.
"""
# Palette: one hue (steel blue), with light-to-dark steps for magnitude
BAR = "#4A6FA5"
INK = "#1d2330"
MUTED = "#5b6474"
GRID = "#e3e8ef"
SEQ_CMAP = plt.cm.Blues  # single-hue sequential ramp


def _short(name, width=48):
    """Clean up a column name for axis labels (remove newlines, shorten)."""
    name = str(name).replace("\n", " ")
    return name if len(name) <= width else name[: width - 1] + "…"


def plot_missing_bar(missing_df, n_rows, show_complete=False, save_path=None):
    """Ranked horizontal bar chart built from the missing_df."""
    summary = missing_df.sort_values("Missing %", ascending=False)
    if not show_complete:
        summary = summary[summary["Missing Count"] > 0]

    # Plot in ascending order so the largest bar ends up at the top
    summary = summary.iloc[::-1]
    labels = [_short(c) for c in summary.index]

    fig, ax = plt.subplots(figsize=(10, 0.38 * len(summary) + 1.6))
    bars = ax.barh(labels, summary["Missing %"], color=BAR, height=0.62)

    # Show the value at the end of each bar
    for bar, pct, cnt in zip(bars, summary["Missing %"], summary["Missing Count"]):
        pct_txt = "<0.1%" if 0 < pct < 0.1 else f"{pct:.1f}%"
        ax.text(bar.get_width() + 0.8, bar.get_y() + bar.get_height() / 2,
                f"{pct_txt}  ({int(cnt):,})", va="center", fontsize=9, color=INK,
                bbox=dict(facecolor="white", edgecolor="none", pad=1.5), zorder=3)

    # Reference lines at 10% and 50% (severity thresholds)
    for x, lbl in [(10, "10%"), (50, "50%")]:
        ax.axvline(x, color=MUTED, lw=1, ls="--", zorder=0)
        ax.text(x, 1.01, lbl, ha="center", va="bottom", fontsize=8, color=MUTED,
                transform=ax.get_xaxis_transform())

    ax.set_xlim(0, max(100, summary["Missing %"].max() + 15))
    ax.set_xlabel("Missing values (% of rows)", color=MUTED)
    ax.set_title(f"Missing values by column  (n = {n_rows:,} rows)",
                 loc="left", fontsize=13, fontweight="bold", color=INK, pad=20)

    ax.xaxis.grid(True, color=GRID, lw=0.8)
    ax.set_axisbelow(True)
    for side in ["top", "right", "left"]:
        ax.spines[side].set_visible(False)
    ax.spines["bottom"].set_color(GRID)
    ax.tick_params(axis="y", length=0, labelsize=9, colors=INK)
    ax.tick_params(axis="x", colors=MUTED)

    plt.tight_layout()
    if save_path:
        fig.savefig(save_path, dpi=200, bbox_inches="tight")
    plt.show()
    return fig, ax


def plot_missing_by_year(df, missing_df, year_col="Year", save_path=None):
    """Heatmap of missing % for each column (rows) and year (columns).
    Uses missing_df to pick and order the columns, worst first."""
    order = missing_df[missing_df["Missing Count"] > 0] \
        .sort_values("Missing %", ascending=False).index
    cols = [c for c in order if c != year_col]
    if not cols:
        print("No missing values to plot.")
        return None, None

    grid = df[cols].isna().groupby(df[year_col]).mean().T * 100

    fig, ax = plt.subplots(figsize=(12, 0.4 * len(cols) + 2))
    im = ax.imshow(grid.values, aspect="auto", cmap=SEQ_CMAP, vmin=0, vmax=100)

    ax.set_xticks(range(grid.shape[1]))
    ax.set_xticklabels(grid.columns.astype(int), rotation=45, ha="right", fontsize=8)
    ax.set_yticks(range(grid.shape[0]))
    ax.set_yticklabels([_short(c) for c in grid.index], fontsize=9)

    # Thin white lines between cells
    ax.set_xticks(np.arange(-0.5, grid.shape[1]), minor=True)
    ax.set_yticks(np.arange(-0.5, grid.shape[0]), minor=True)
    ax.grid(which="minor", color="white", lw=1.5)
    ax.tick_params(which="both", length=0, colors=INK)
    for spine in ax.spines.values():
        spine.set_visible(False)

    cbar = fig.colorbar(im, ax=ax, fraction=0.025, pad=0.02)
    cbar.set_label("Missing (% of rows in that year)", color=MUTED)
    cbar.outline.set_visible(False)

    ax.set_title("Missing values by column and year", loc="left",
                 fontsize=13, fontweight="bold", color=INK)
    plt.tight_layout()
    if save_path:
        fig.savefig(save_path, dpi=200, bbox_inches="tight")
    plt.show()
    return fig, ax
```

```python
plot_missing_bar(eda_results, n_rows=len(df), save_path=IMAGES / "missing_by_column.png")
plot_missing_by_year(df, eda_results, year_col="Year", save_path=IMAGES / "missing_by_year.png")
```

<h3 style="color: #1F4E79;">4.3. Skewness analysis: </h3>

```python
"""
Skewness analysis for the numeric columns in df.

Uses the DataFrame `df` that is already loaded in the notebook.
Outputs:
  - skew_df : table with skewness, mean, median, mean/median ratio and a label
  - a ranked bar chart of skewness
"""
# Columns to leave out: identifiers and coordinates (their skew has no meaning)
EXCLUDE = ["Year", "Latitude", "Longitude"]

INK, MUTED, GRID = "#1d2330", "#5b6474", "#e3e8ef"
RIGHT, LEFT = "#4A6FA5", "#C26A3D"   # right skew = blue, left skew = orange


def skew_label(s):
    """Common rule of thumb for interpreting skewness."""
    a = abs(s)
    if a < 0.5:
        return "Approximately symmetric"
    side = "right" if s > 0 else "left"
    if a < 1:
        return f"Moderately {side}-skewed"
    return f"Highly {side}-skewed"


def compute_skewness(df, exclude=EXCLUDE):
    """Return a skewness table for all numeric columns."""
    data = df.copy()

    # Density is stored as text (e.g. "1,265"), so convert it to a number first
    for c in data.columns:
        if "Density" in str(c) and not pd.api.types.is_numeric_dtype(data[c]):
            data[c] = pd.to_numeric(data[c].astype(str).str.replace(",", ""),
                                    errors="coerce")

    num = data.select_dtypes(include="number").drop(columns=exclude, errors="ignore")

    skew_df = pd.DataFrame({
        "Skewness": num.skew(),          # pandas skips NaN by default
        "Mean": num.mean(),
        "Median": num.median(),
    })
    skew_df["Mean / Median"] = np.where(skew_df["Median"] > 0,
                                        skew_df["Mean"] / skew_df["Median"], np.nan)
    skew_df["Interpretation"] = skew_df["Skewness"].apply(skew_label)
    skew_df["Suggest log transform"] = (skew_df["Skewness"] > 1) & (num.min() >= 0)
    return skew_df.sort_values("Skewness", ascending=False)


def plot_skewness(skew_df, save_path=None):
    """Ranked horizontal bar chart of skewness, with ±0.5 and ±1 guide lines."""
    s = skew_df["Skewness"].iloc[::-1]   # largest at the top
    labels = [str(c).replace("\n", " ")[:48] for c in s.index]
    colors = [RIGHT if v >= 0 else LEFT for v in s]

    fig, ax = plt.subplots(figsize=(10, 0.38 * len(s) + 1.6))
    bars = ax.barh(labels, s, color=colors, height=0.62)

    # Value labels sit just right of the bar (or of zero for negative bars)
    pad = 0.012 * (s.max() - min(s.min(), 0) + 3)
    for bar, v in zip(bars, s):
        ax.text(max(v, 0) + pad, bar.get_y() + bar.get_height() / 2,
                f"{v:.2f}", va="center", ha="left", fontsize=9, color=INK,
                bbox=dict(facecolor="white", edgecolor="none", pad=1.5), zorder=3)

    for x in [-1, -0.5, 0.5, 1]:
        ax.axvline(x, color=MUTED, lw=1, ls="--", zorder=0)
    ax.axvline(0, color=INK, lw=1)

    ax.set_xlim(min(-1.5, s.min() - 1), s.max() + 1.5)
    ax.set_xlabel("Skewness  (|s| < 0.5 symmetric · 0.5–1 moderate · > 1 high)",
                  color=MUTED)
    ax.set_title("Skewness by column", loc="left", fontsize=13,
                 fontweight="bold", color=INK, pad=12)

    ax.xaxis.grid(True, color=GRID, lw=0.8)
    ax.set_axisbelow(True)
    for side in ["top", "right", "left"]:
        ax.spines[side].set_visible(False)
    ax.spines["bottom"].set_color(GRID)
    ax.tick_params(axis="y", length=0, labelsize=9, colors=INK)
    ax.tick_params(axis="x", colors=MUTED)

    plt.tight_layout()
    if save_path:
        fig.savefig(save_path, dpi=200, bbox_inches="tight")
    plt.show()
    return fig, ax


# Run on the existing df
skew_df = compute_skewness(df)
print("=== SKEWNESS ===")
print(skew_df.round(2).to_string())
plot_skewness(skew_df, save_path=IMAGES / "skewness_by_column.png")
```

<h3 style="color: #1F4E79;">4.9. Fills gaps using facts and definitions, not estimates:</h3>

<h4> Structural fixes (deterministic, not statistical) </h4>

**Static geography.** Land area, coordinates and density are one value per country in this dataset. A single missing cell can be copied from the same country's other rows with zero risk.

**Nuclear from an accounting identity.** Low-carbon share is defined as
$LC = \dfrac{N + R}{F + N + R}$ (as a fraction). Solving for nuclear gives
$N = \dfrac{LC\,(F + R) - R}{1 - LC}$.
So when fossil, renewables and low-carbon share are known, nuclear is *computed*, not guessed. Remaining gaps get 0 only for countries that never report any nuclear output.

1. Static geography. Land area, latitude, longitude and population density have one fixed value per country in this dataset. They are the same in every year. So if one row is missing a value, the correct answer is already in that country’s other rows, and copying it in introduces no error.

2. Nuclear from an accounting identity. Low-carbon electricity is nuclear plus renewables, as a share of total generation:

LC = (N + R) / (F + N + R)

Here F is fossil, N is nuclear and R is renewables, all in TWh, and LC is a fraction from 0 to 1. If you know LC, F and R, you can solve for N:

LC·(F + N + R) = N + R
LC·F + LC·R − R = N − LC·N
N = (LC·(F + R) − R) / (1 − LC)

The algebra is correct. In other words, nuclear output is already implied by the other three columns. Any gaps still left are filled with 0, but only for countries that never report nuclear output.

```python
STATIC = ["density_p_km2", "land_area_km2", "lat", "lon"]               # columns that don't change over time
df[STATIC] = df.groupby("country")[STATIC].transform(lambda s: s.ffill().bfill())  # copy each country's own values through ffill() and bfill() into its gaps

lc = df["low_carbon_elec_pct"] / 100                                    # low-carbon share as a fraction (0–1)
F, R = df["elec_fossil_twh"], df["elec_ren_twh"]                       # short aliases for readability
nuc_calc = (lc * (F + R) - R) / (1 - lc)                                # nuclear implied by the identity above
can_calc = df["elec_nuclear_twh"].isna() & lc.lt(1) & F.notna() & R.notna()  # only where inputs exist and LC < 100 %
df.loc[can_calc, "elec_nuclear_twh"] = nuc_calc[can_calc].clip(lower=0).round(2)  # fill, never below zero
print("Nuclear values recovered from the identity:", int(can_calc.sum()))

never_nuclear = df.groupby("country")["elec_nuclear_twh"].transform(lambda s: s.fillna(0).max() == 0)  # True if country never reports >0
fill_zero = df["elec_nuclear_twh"].isna() & never_nuclear               # missing AND the country has no nuclear history
df.loc[fill_zero, "elec_nuclear_twh"] = 0.0                             # no reactors -> zero nuclear output
print("Nuclear values set to 0 (country never had nuclear):", int(fill_zero.sum()))
```

*Static geography* fills the single missing value in density_p_km2, land_area_km2, lat and lon. It copies the value from the same country’s other rows: first from earlier years (ffill), then from later years (bfill). A country with only one row, such as French Guiana, has nothing to copy from, so its gap stays NaN.

*Nuclear from the identity* calculates missing elec_nuclear_twh values from the low-carbon identity. It only does this when fossil, renewables and low-carbon share are all present, and when low-carbon share is below 100%, to avoid dividing by zero. Small negative results caused by rounding are clipped to 0.

*Zero for countries without nuclear power*, that is  if a country never reports any nuclear output, its remaining gaps are set to 0. Countries with nuclear output in any year are skipped, so their gaps are left for later steps.

<h4> Within-country time interpolation </h4>

Energy indicators move slowly, so a gap between two observed years is well approximated by a straight line between them. Two safeguards:

* `method="index"` with **year as the index** — a gap from 2004 to 2008 is spread over 4 years correctly even if rows are missing.
* `limit_area="inside"` — we only fill **between** observed values. No extrapolation past a country's first or last observation. That is what keeps 2020 CO₂ empty and prevents future information leaking backwards into a forecasting model.

`gdp_growth_pct` is **not** interpolated: growth is shock-driven (COVID, conflict), and a smoothed value would erase exactly the signal we care about.

```python
before = df[IMPUTE_COLS].isna().sum()                                   # missing counts before interpolation

tmp = df.set_index("year")                                              # make year the index so gaps are measured in years
filled = (tmp.groupby("country")[IMPUTE_COLS]                           # treat each country separately
             .transform(lambda s: s.interpolate(method="index",          # straight line between known years
                                                limit_area="inside")))   # never extend past the first/last observation
df[IMPUTE_COLS] = filled.to_numpy()                                     # write back (row order is unchanged)

after = df[IMPUTE_COLS].isna().sum()                                    # missing counts after interpolation
print(pd.DataFrame({"missing_before": before, "missing_after": after, "filled": before - after}).to_string())
```

*Interpolation filled nothing.* The check below counts **interior gaps**: missing values with observed data on both sides. If every count is 0, all remaining gaps sit at the start or end of a country's series (or cover the whole series), so `limit_area="inside"` correctly leaves them empty.

```python
# Count interior gaps: missing values with an observed value both before and after (within the same country)
interior = (df.groupby("country")[IMPUTE_COLS]
             .transform(lambda s: s.isna() & s.ffill().notna() & s.bfill().notna())
             .sum())
print("Interior gaps per column:")
print(interior.to_string())
```

<h3 style="color: #1F4E79;">4.10. Recomputed low-carbon share where the generation mix is complete:</h3>

If fossil, nuclear and renewable generation are all known, the low-carbon share is a definition, not an estimate.

```python
total_gen = df["elec_fossil_twh"] + df["elec_nuclear_twh"] + df["elec_ren_twh"]   # total generation (TWh)
fixable = df["low_carbon_elec_pct"].isna() & total_gen.gt(0)                       # missing share but a usable denominator
df.loc[fixable, "low_carbon_elec_pct"] = (                                          # fill from the definition
    (df["elec_nuclear_twh"] + df["elec_ren_twh"]) / total_gen * 100)[fixable]
print("Low-carbon shares recomputed from the mix:", int(fixable.sum()))
```

<h3 style="color: #1F4E79;">4.11. Outlier flagged (not removed):</h3>

We use a *robust z-score* based on the median and the median absolute deviation (MAD) instead of the mean and standard deviation because the mean and SD are themselves dragged around by the outliers we are trying to find. |z| > 5 is a deliberately strict cut-off so only genuine shocks are flagged.

```python
g = df["gdp_growth_pct"]                                                # alias for readability
med = g.median()                                                        # robust centre
mad = (g - med).abs().median()                                          # robust spread
df["gdp_growth_robust_z"] = 0.6745 * (g - med) / mad                    # 0.6745 rescales MAD to match an SD for normal data
df["gdp_growth_outlier"] = (df["gdp_growth_robust_z"].abs() > 5).astype(int)  # 1 = extreme growth year

print("Flagged extreme GDP-growth years:", int(df["gdp_growth_outlier"].sum()))
print(df.loc[df["gdp_growth_outlier"] == 1, ["country", "year", "gdp_growth_pct"]]
        .sort_values("gdp_growth_pct").to_string(index=False))
```

<h3 style="color: #1F4E79;">4.12. Country data-coverage score:</h3>

Low coverage is driven mainly by country-name mismatches in the source data (for example, Egypt vs. Egypt, Arab Rep.) and by overseas territories, not by development level. A coverage score lets results be reported with and without these countries.

```python
CORE = ["elec_access_pct", "clean_cooking_pct", "ren_share_tfec_pct", "low_carbon_elec_pct",
        "primary_energy_kwh_pc", "energy_intensity_mj_usd", "co2_kt", "gdp_pc_usd", "gdp_growth_pct"]  # core indicators

coverage = raw.groupby("country")[CORE].apply(lambda g: g.notna().mean().mean())   # share of ORIGINAL cells present per country
df["coverage_score"] = df["country"].map(coverage).round(3)                         # attach the score to every row
df["low_coverage"] = (df["coverage_score"] < 0.70).astype(int)                     # 1 = less than 70 % of core data was observed

print("Countries with low coverage (< 70 % of core cells observed):")
print(coverage[coverage < 0.70].sort_values().round(2).to_string())
```
